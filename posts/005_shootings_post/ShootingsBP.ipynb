# **Top Gun Shot Wounds in Philadelphia Across Data Sets**

### While exploring this data, one of the most prominent points was wounds being reported. The two data sets show similar numbers for wound reporting, althouht they are slightly different.

### Looking at this graph, it is clear that "multiple" and "leg" have the highest percentage differences. This could be for many reasons. "Multiple" makes sense in that it can be hard to keep track of multiple elements to a patient. "Leg" most likely could have such a high differnece as well because the leg is a common place to shoot a victim without killing them, so they are immobile. "Head" and "Chest" have a lower percentage difference likely because those are fatal shots. It is more clear to people who collect this data when these wounds are present because they leave such an impact. 

### This shows that perhaps when it comes to data collection, the observations being easier to observe may leave an impact on the data being collected. As a result, more visible or obvious characteristics may be recorded more consistently, while less clear details are subject to variation or omission. This can ultimately shape the patterns we see in the data, influencing how we interpret and respond to these events.

# **Reporting the Sex of Victims Between Two Data Sets**

### To further examine the consistency between the two Philadelphia shooting datasets, I analyzed the distribution of sex among individuals involved in the incidents. Sex is a demographic variable that is typically straightforward to observe and record, making it a useful benchmark for evaluating data reliability. If both datasets are capturing similar populations, we would expect the proportions of male and female individuals to be relatively consistent across sources. By comparing these distributions, we can assess whether both datasets are aligned in their basic demographic reporting.


![](sex.png){width=500}

### The comparison of sex across the two datasets reveals a noticeable difference in how male and female individuals are represented. While both datasets show that males make up the majority of those involved in shootings, the gap in reported percentages is larger for men than for women. This indicates that the datasets differ more in how they capture male involvement, even though the overall pattern remains consistent across both sources.

### However, this difference is likely influenced by the fact that men are far more frequently involved in shootings overall. Because the base number of male cases is much higher, even small variations in reporting or classification can lead to larger absolute differences. In contrast, the number of female cases is relatively smaller, so differences between datasets appear less pronounced. This suggests that the observed discrepancy may not necessarily reflect a meaningful inconsistency, but rather the underlying distribution of the data itself.

# **Outside or Inside Reporting Diefference Between the Data Sets**

### Looking at whether shootings occurred inside or outside adds another layer to understanding how these datasets compare. Unlike something like sex, which is usually straightforward to record, location context can be a bit more subjective and dependent on how the incident is reported. By comparing the share of inside versus outside shootings across both datasets, we can see how consistently this type of detail is captured.

![](inside.png){width=500}

### The graph shows a larger difference between the datasets for outside shootings than for inside ones. At first, this might suggest inconsistency in how outside incidents are recorded, but it is more likely a reflection of the fact that most shootings occur outside. Because outside cases make up a much larger portion of the data, even small differences in how they are classified can lead to more noticeable gaps between datasets. In contrast, inside shootings are less common, so any variation appears smaller.

# **Conclusion:**

### Overall, this analysis reveals that there are a wide variety of inconsistencies between the two datasets, even though they are describing the same type of events in the same city. While both datasets generally follow similar patterns, the differences in how certain variables are recorded show that the data is not perfectly aligned. 

### These inconsistencies are especially noticeable in categories with a larger number of observations, such as males or outside shootings. Because these groups make up a bigger portion of the data, even small differences in reporting or classification can appear more pronounced. 

### In contrast, categories with fewer observations tend to show smaller differences, not necessarily because they are more accurate, but because there is less data where discrepancies can occur. **This highlights the importance of recognizing how both data collection methods and underlying distributions can shape the patterns we see in the data.**
