```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
mbti_df = pd.read_csv('mbti.csv')
```

```python
nltk.download('vader_lexicon')
```

```python
extrovert = mbti_df[mbti_df["type"].str.startswith("E")]
```

```python
sid = SentimentIntensityAnalyzer()
```

```python
extrovert_compound_mean = extrovert["posts"].astype(str).apply(lambda x: sid.polarity_scores(x)["compound"]).mean()

print(extrovert_compound_mean)
```

```python
extrovert_compound_min = extrovert["posts"].astype(str).apply(lambda x: sid.polarity_scores(x)["compound"]).min()

print(extrovert_compound_min)
```

```python
extrovert_compound_max = extrovert["posts"].astype(str).apply(lambda x: sid.polarity_scores(x)["compound"]).max()

print(extrovert_compound_max)
```

```python
introvert = mbti_df[mbti_df["type"].str.startswith("I")]
print(introvert)
```
