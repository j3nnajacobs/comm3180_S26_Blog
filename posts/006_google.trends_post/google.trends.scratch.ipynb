```python
quizlet_df = pd.read_csv('quizlet.csv')
coolmath_df = pd.read_csv('cool.math.games.csv')
```

## Merge data set and clean 

```python
merged_df = pd.merge(quizlet_df, coolmath_df)

print(merged_df.head())
```

## Correlation: When quizlet increase, cool math games increases

```python
correlation = merged_df["Quizlet"].corr(merged_df["Cool Math Games"])

print("Correlation:", correlation)
```

### A lot of people would expect:studying goes up, games go down

But data suggests students may be using both more during the same periods.
That supports a diff interpretation:
 - periods of high academic engagement may also increase procrastination or stress-relief behavior online.
   
That’s a more interesting behavioral insight.

## BUT, correlation is not greatly strong, just moderate. 

## WHY IS THAT? Going to look at the rise and fall of the

```python
quizlet = quizlet.rename(columns={"Time": "date", "Quizlet": "quizlet"})
coolmath = coolmath.rename(columns={"Time": "date", "Cool Math Games": "cool_math_games"})

quizlet["date"] = pd.to_datetime(quizlet["date"])
coolmath["date"] = pd.to_datetime(coolmath["date"])

merged_df = pd.merge(quizlet, coolmath, on="date", how="inner")

merged_df.head()
```

```python
merged_df["month"] = merged_df["date"].dt.month
merged_df["month_name"] = merged_df["date"].dt.month_name()

monthly_avg = merged_df.groupby("month_name")[["quizlet", "cool_math_games"]].mean()

month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

monthly_avg = monthly_avg.reindex(month_order)

monthly_avg
```

```python
monthly_avg["cool_math_games"].plot(kind="bar", figsize=(10,6), color=["green"])

plt.title("Average Cool Math Games Search Interest by Month")
plt.xlabel("Month")
plt.ylabel("Average Search Interest")
plt.xticks(rotation=45)
plt.show()
```

```python
def school_phase(month):
    if month in [8, 9]:
        return "Start of School Year"
    elif month in [10, 11]:
        return "Mid Semester"
    elif month == 12:
        return "Finals / End Semester"
    elif month in [6, 7]:
        return "Summer"
    elif month in [3, 5]:
        return "Other School Months"
    elif month in [1, 2]:
        return "Post Winter Break Months"

merged_df["school_phase"] = merged_df["month"].apply(school_phase)

phase_avg = merged_df.groupby("school_phase")[["quizlet", "cool_math_games"]].mean()

phase_avg
```

```python
phase_avg["cool_math_games"].plot(kind="bar", figsize=(8,5))

plt.title("Cool Math Games Search Interest by School Phase")
plt.xlabel("School Phase")
plt.ylabel("Average Search Interest")
plt.xticks(rotation=45)
plt.show()
```

```python
monthly_avg["quizlet"].plot(kind="bar", figsize=(10,6))

plt.title("Average Quizlet Search Interest by Month")
plt.xlabel("Month")
plt.ylabel("Average Search Interest")
plt.xticks(rotation=45)

plt.show()
```

```python
phase_avg["quizlet"].plot(kind="bar", figsize=(8,5))

plt.title("Quizlet Search Interest by School Phase")
plt.xlabel("School Phase")
plt.ylabel("Average Search Interest")
plt.xticks(rotation=30)

plt.show()
```
