# **Introduction:**

### Crime in cities is often discussed in terms of where it happens, with certain neighborhoods gaining reputations for higher levels of criminal activity. However despite this observation, people tend to forget about another relevant aspect-- Time. 

### Throughout a typical day, the rhythms of daily life shift as people go to work, go out at night, and move through public spaces. Crime statistics don't just change by location, but by the hour of the day. 

### Because of this, it’s reasonable to expect that different types of crime may occur at different times of day. A crowded afternoon in a busy commercial area may create opportunities for theft, while late-night hours might see more conflicts between individuals. When you start to look at crime data through the filter of time, patterns can begin to emerge that are not obvious at first glance.  

### This raises an interesting question: **do certain types of crime tend to occur at specific hours of the day?** Using Philadelphia police incident data from 2025, I analyze when different crimes are most likely to happen by examining the most common offenses reported during each hour of the day. By exploring how crime patterns shift over a 24-hour period, we can **better understand how daily human activity shapes the types of incidents that occur across a city.**

# **Rise and Fall of Top 4 Crimes in Philadelphia**

### While the previous chart showed how total crime fluctuates throughout the day, it does not reveal which specific offenses are driving these patterns. To explore this further, I focused on the four most common crimes in the dataset and examined how frequently each occurs during every hour of the day. Plotting these trends side by side provides a clearer picture of **how different types of crime follow distinct daily patterns.**

![](24.hr.trend.png){width=500}

### Several interesting trends emerge from this visualization. Thefts (shown in orange) occur far more frequently than the other crimes and show a noticeable increase beginning in the late morning, peaking during the afternoon around 4–5 PM. **This pattern likely reflects times when more people are out in public spaces, creating greater opportunities for theft.** Other assaults (purple) also rise throughout the day but tend to remain relatively high into the evening hours, suggesting that interpersonal conflicts may be more common later in the day when social activity increases. There could  likely be a link to alcohol consumption increasing at night. A research study by a National Health and Nutrition Examination Survey found that "young adults are exposed to alcohol and its acute effects at a much higher level during the night, and this in-turn increases the risk of alcohol-related injuries during that time." showing that this could be a heavy cause for the crime rate as well.*

# **Conclusion:**

### Overall, this analysis highlights that crime patterns are closely tied to the rhythms of daily life in an urban city. Examining both overall crime levels and the timing of specific offenses, it becomes clear that different types of crime follow distinct patterns throughout the day. Theft, assault, and vehicle theft each peak at different hours, suggesting that criminal activity is influenced by when people are most active in public spaces. Rather than crime occurring randomly across time, the data reveals predictable trends tied to how and when people move through urban environments.

### In Philadelphia, these patterns reflect broader relationships between social routines and crime. Cities around the world experience similar daily cycles as people commute to work, gather socially, and return home at night. These shifts in activity create changing opportunities for different types of crime to occur. Understanding when crimes are most likely to happen can help policymakers and law enforcement better allocate resources and develop  prevention strategies.
