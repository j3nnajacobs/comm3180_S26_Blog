```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
athlete_df = pd.read_csv("bios.csv")
```

```python
uniq_athlete = athlete_df.dropna(subset=["born_country", "NOC"])
```

```python
noc_to_country = {
    "France": "FRA",
    "United States": "USA",
    "Canada": "CAN",
    "Germany": "GER",
    "Russia": "RUS",
    "Italy": "ITA",
    "Switzerland": "SUI",
    "Austria": "AUT",
    "Sweden": "SWE",
    "Norway": "NOR",
    "Finland": "FIN",
    "Netherlands": "NED",
    "China": "CHN",
    "Japan": "JPN",
    "South Korea": "KOR",
    "Czech Republic": "CZE",
    "Slovakia": "SVK",
    "Poland": "POL",
    "Ukraine": "UKR",
    "Belarus": "BLR",
    "Morroco" : "MAR",
    "Great Britain" : "GBR",
    "Iraq" : "IRQ"
}
```

```python
uniq_athlete["NOC_country"] = uniq_athlete["NOC"].replace(noc_to_country)
```

```python
switch_athletes = switch_athletes.dropna(subset=["NOC_country"])
```

```python
switch_athletes = uniq_athlete[uniq_athlete["born_country"] != uniq_athlete["NOC_country"]]
```

```python
switch_athletes[["name", "born_country", "NOC_country"]].head()
```

```python
amt_of_switch = len(switch_athletes)
print(amt_of_switch)
```

```python
len(uniq_athlete)
```

```python
percent_of_switch = (amt_of_switch) / len(uniq_athlete)
print(percent_of_switch)
```

```python
top_10_import = switch_athletes["NOC_country"].value_counts().head(10)
print(top_10_import)

##Top countries that have atheltes that were born in diff country
```

```python
plt.bar(top_10_import.index, top_10_import.values, color="lightblue")

plt.title("Top 10 Countries Gaining Olympic Talent From Foreign Countries")
plt.xlabel("Country Represented")
plt.ylabel("Number of Foreign-Born Athletes")

plt.xticks(rotation=90)

plt.show()
```

```python
top_10_export = switch_athletes["born_country"].value_counts().head(10)
switch_athletes["born_country"].value_counts().head(10)

## top countries people are leaving 
```

```python
plt.bar(top_10_export.index, top_10_export.values, color="red")

plt.title("Top 10 Countries Losing Their Winter Olympic Talent")
plt.xlabel("Countries Athletes Are Leaving")
plt.ylabel("Number of Exported Atheles")

plt.xticks(rotation=90)

plt.show()
```
